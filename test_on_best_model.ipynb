```python
import pickle
import pandas as pd

# 1. Load the Netflix model
model_path = r"C:\Users\Royal\files3\best_model.pkl"
with open(model_path, 'rb') as file:
    model = pickle.load(file)
print("1. Netflix Model Loaded Successfully!")

# 2. Load the PROCESSED Netflix dataset
# Changed from netflix_titles_CLEANED.csv to netflix_processed.csv
netflix_data = pd.read_csv(r"C:\Users\Royal\files3\netflix_processed.csv")

# 3. Take 5 random data samples
random_samples = netflix_data.sample(n=5)
print("2. Selected 5 Random Processed Samples.")

# 4. Automatically select only the features the model was trained on
expected_features = model.feature_names_in_
X_random = random_samples[expected_features]

# 5. Apply the model to predict on the random samples
predictions = model.predict(X_random)

# 6. Display final results in a structured format
results = pd.DataFrame({
    'Sample_Index': random_samples.index,
    'Predicted_Value': predictions
})

print("\nModel Predictions for Random Samples:")
print(results)
```
